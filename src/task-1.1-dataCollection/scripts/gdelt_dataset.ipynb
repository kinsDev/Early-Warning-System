```python
import gdelt
import pandas as pd

# Initialize GDELT 2.0
gd2 = gdelt.gdelt(version=2)

# Define the date range for querying (expand this range if necessary)
date_range = ['2024 10 01', '2024 10 03']  # Adjust dates as needed

# Pull data for the specified dates, covering all 15-minute intervals
results = gd2.Search(date_range, table='events', coverage=True)

# Check if results are empty
if len(results) == 0:
    print("No results found.")
else:
    # Convert results to a pandas DataFrame
    df = pd.DataFrame(results)
    
    # Display basic info about the DataFrame
    print(f"Total records found: {len(df)}")
    print("DataFrame columns:", df.columns.tolist())

    # Optionally filter for specific event types related to humanitarian crises
    # Example: Using a broader range of event codes for testing
    # humanitarian_df = df[df['EventCode'].isin([190, 175, 14, 178])]  # Uncomment to filter
    humanitarian_df = df  # For testing, using all records

    # Print the length of the filtered DataFrame and display the first few entries
    print(f"Number of records after filtering: {len(humanitarian_df)}")
    print(humanitarian_df.head())

  
    humanitarian_df.to_csv('humanitarian_crises_data.csv', index=False)
    print("Data saved to humanitarian_crises_data.csv")

```

```python
import pandas as pd

# Load the large CSV file
file_path = "humanitarian_crises_data_october_1-3.csv"
df = pd.read_csv(file_path)

# Calculate the number of rows to split into 5 equal parts
total_rows = len(df)
rows_per_part = total_rows // 5

# Split the data into five parts
df_part1 = df.iloc[:rows_per_part]
df_part2 = df.iloc[rows_per_part:2*rows_per_part]
df_part3 = df.iloc[2*rows_per_part:3*rows_per_part]
df_part4 = df.iloc[3*rows_per_part:4*rows_per_part]
df_part5 = df.iloc[4*rows_per_part:]

# Save the five parts as separate CSV files
df_part1.to_csv("humanitarian_crises_data_part1.csv", index=False)
df_part2.to_csv("humanitarian_crises_data_part2.csv", index=False)
df_part3.to_csv("humanitarian_crises_data_part3.csv", index=False)
df_part4.to_csv("humanitarian_crises_data_part4.csv", index=False)
df_part5.to_csv("humanitarian_crises_data_part5.csv", index=False)

print(f"CSV file split into five parts, each containing around {rows_per_part} rows.")
```
