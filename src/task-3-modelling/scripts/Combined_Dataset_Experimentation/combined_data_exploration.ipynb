```python
# Function to get unique values and their data types for each column
def get_unique_values_with_dtypes(data, columns):
    result = []
    for column in columns:
        unique_values = data[column].unique()
        dtypes = set(type(val) for val in unique_values)
        result.append({
            'Column': column,
            'Unique Values': unique_values,
            'Data Types': dtypes
        })
    return pd.DataFrame(result)

# Filter the data to include only the matched columns
existing_columns = [col for col in selected_columns if col in data.columns]
data_filtered = data[existing_columns]

# Get unique values and their data types for each column
unique_values_df = get_unique_values_with_dtypes(data_filtered, existing_columns)

# Display the result
print(unique_values_df)
```
